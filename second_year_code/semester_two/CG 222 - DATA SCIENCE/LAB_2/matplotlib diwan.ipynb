```
import pandas as pd

data = {
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Math": [85, 78, 92, 70, 88, 95],
    "Science": [80, 85, 90, 75, 84, 98],
    "Hours_Studied": [5, 4, 6, 3, 5, 7]
}

data = pd.DataFrame(data)

```

```
data.head()
```

```
df = data
```

#Line Plot

```
import matplotlib.pyplot as plt

plt.plot(df["Student"], df["Math"], marker="o")

plt.title("Math Scores")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

#Bar Chart
Compare categories

```
plt.bar(df["Student"], df["Math"])

plt.title("Math Scores")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

#Horizontal Bar Chart

```
plt.barh(df["Student"], df["Science"])

plt.title("Science Scores")

plt.show()
```

#Scatter Plot
Relationship between study hours and marks

```
plt.scatter(df["Hours_Studied"],
            df["Math"])

plt.title("Study Hours vs Math Score")
plt.xlabel("Hours Studied")
plt.ylabel("Math Marks")

plt.show()
```

#Histogram
Distribution of marks.

```
plt.hist(df["Math"],
         bins=5)

plt.title("Distribution of Math Scores")

plt.show()
```

#Pie Chart
Percentage contribution of total math marks.

```
plt.pie(
    df["Math"],
    labels=df["Student"],
    autopct="%1.1f%%"
)

plt.title("Math Score Contribution")

plt.show()
```

#Box Plot
Detect spread and outliers.

```
plt.boxplot(df["Math"])

plt.title("Math Scores Box Plot")

plt.show()
```

#Multiple Line Graphs
Compare Math and Science scores.

```
plt.plot(df["Student"],
         df["Math"],
         marker="o",
         label="Math")

plt.plot(df["Student"],
         df["Science"],
         marker="s",
         label="Science")

plt.legend()

plt.title("Math vs Science")

plt.show()
```

#Subplots
Display multiple graphs in one figure.

```
fig, ax = plt.subplots(1, 2, figsize=(10,4))

ax[0].bar(df["Student"], df["Math"])
ax[0].set_title("Math")

ax[1].bar(df["Student"], df["Science"])
ax[1].set_title("Science")

plt.tight_layout()
plt.show()
```

#Stack Plot
Show combined subject performance.

```
x = range(len(df))

plt.stackplot(
    x,
    df["Math"],
    df["Science"],
    labels=["Math", "Science"]
)

plt.legend()
plt.title("Stack Plot")

plt.show()
```

#Area Plot

```
plt.fill_between(
    df["Student"],
    df["Math"],
    alpha=0.5
)

plt.title("Math Area Plot")

plt.show()
```

#Stem Plot

```
plt.stem(
    df["Hours_Studied"],
    df["Math"]
)

plt.title("Stem Plot")

plt.show()
```

#Animation Graph (Most Important)
Animated Growing Line Chart

```
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

x = []
y = []

fig, ax = plt.subplots()

students = ["A", "B", "C", "D", "E", "F"]
scores = [85, 78, 92, 70, 88, 95]

def animate(i):
    x.append(students[i])
    y.append(scores[i])

    ax.clear()
    ax.plot(x, y, marker='o')

    ax.set_title("Animated Math Scores")
    ax.set_xlabel("Students")
    ax.set_ylabel("Marks")

ani = FuncAnimation(
    fig,
    animate,
    frames=len(students),
    interval=1000,
    repeat=False
)

HTML(ani.to_jshtml())
```

How it Works
- FuncAnimation() repeatedly calls animate()
- frames=6 → runs 6 times
- interval=1000 → updates every 1 second
- Each frame adds one new data point

#Real-Time Random Data Animation
This simulates:
- Stock prices
- Sensor data
- IoT dashboards
- CPU monitoring

```
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import random

fig, ax = plt.subplots(figsize=(8,4))

x = []
y = []

line, = ax.plot([], [], marker='o')

ax.set_xlim(0, 100)
ax.set_ylim(50, 110)

ax.set_title("Live Score Stream")
ax.set_xlabel("Frame")
ax.set_ylabel("Score")

def animate(i):
    x.append(i)
    y.append(random.randint(60,100))

    line.set_data(x, y)
    return line,

ani = FuncAnimation(
    fig,
    animate,
    frames=100,
    interval=100,
    blit=True,
    cache_frame_data=False
)

HTML(ani.to_jshtml())
```
